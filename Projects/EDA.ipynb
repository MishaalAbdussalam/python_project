```python
# Asia has the largest world population percentage of 59% followed by Africa with 17%. Oceania(Australia i suppose) has the least world population percentage of 0.55%.
df.groupby('Continent').sum(numeric_only=True).sort_values(by = 'World Population Percentage', ascending = False)
```

```python
population_columns = [
    '1970 Population',
    '1980 Population',
    '1990 Population',
    '2000 Population',
    '2010 Population',
    '2015 Population',
    '2020 Population',
    '2022 Population']
df2 = df.groupby('Continent')[population_columns].mean(numeric_only=True).sort_values(by = '2022 Population', ascending = False)
df2
```

```python
df3 = df2.transpose()
df3
```

```python
df3.plot(figsize = (12,6))
```

```python
# Here in 1970 asia is already in the lead by quite a margin and the gap increases drastically in the 2000s and then levels of at around 2020.
```
